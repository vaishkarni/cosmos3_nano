# An Introduction to Cosmos 3

Cosmos 3 is an open omni-model for Physical AI: a single model that reasons about, generates, and acts in the physical world. In this course you will work hands-on across its three core capabilities — **reasoning**, **world generation**, and **action prediction** — and see how they connect into one synthetic-data pipeline for robotics, autonomous vehicles, and smart infrastructure.

### 3. Getting access to gated models

The World Generation and Action notebooks run Cosmos 3 through the Cosmos Framework, which downloads model weights from Hugging Face the first time you run inference. These repositories are **gated** and subject to the [*NVIDIA Open Model License*](https://www.nvidia.com/en-us/agreements/enterprise-software/nvidia-open-model-license/), so you must accept their license agreements before the download will succeed.

Follow the links below and request access to each:
- [https://huggingface.co/nvidia/Cosmos3-Nano](https://huggingface.co/nvidia/Cosmos3-Nano) — the Cosmos 3 Nano omni-model (single-GPU) used for generation **and** all three action modes (forward dynamics, inverse dynamics, policy).
- [https://huggingface.co/nvidia/Cosmos-Guardrail1](https://huggingface.co/nvidia/Cosmos-Guardrail1) — the safety guardrail the generator enables by default.

> The larger `nvidia/Cosmos3-Super` checkpoint is optional and only needed if you choose to run the multi-GPU Super examples.

> [`nvidia/Cosmos3-Nano-Policy-DROID`](https://huggingface.co/nvidia/Cosmos3-Nano-Policy-DROID) is optional — accept it only for the advanced, out-of-lab *closed-loop* RoboLab policy exercise at the end of Part 4. The in-notebook policy run uses `Cosmos3-Nano` above.

> The World Reasoning notebook uses the Cosmos 3 Reasoner **NIM** instead, which authenticates with an NGC key (handled by the lab environment), so it does not require Hugging Face access.

Follow these steps to accept each license agreement:
